```python
import pandas as pd

dfs = {
    "VAR": pd.read_csv("ta_var.tsv", sep="\t").drop(columns=["distractor_match_rate"]),
    "HOM": pd.read_csv("ta_hom.tsv", sep="\t"),
    "REL": pd.read_csv("ta_rel.tsv", sep="\t"),
}

ta = pd.concat(
    [df.set_index("model").round(1) for df in dfs.values()],
    axis=1,
    keys=dfs.keys()
).reset_index()
```

```python
ta
```

```python
ta.to_csv("2_ta.tsv", sep="\t", index=False)
```
